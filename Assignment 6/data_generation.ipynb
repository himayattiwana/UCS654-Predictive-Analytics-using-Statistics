# Data Generation using Physics-Based Simulation
This notebook generates a synthetic dataset using projectile motion equations.

```python
import numpy as np
import pandas as pd
```

```python
def projectile_range(v, theta, g=9.81):
    theta_rad = np.radians(theta)
    return (v**2 * np.sin(2 * theta_rad)) / g
```

```python
def generate_dataset(n_samples=1000):
    velocities = np.random.uniform(12, 95, n_samples)
    angles = np.random.uniform(15, 88, n_samples)
    true_ranges = projectile_range(velocities, angles)
    noise = np.random.normal(0, 25, n_samples)
    measured_ranges = true_ranges + noise
    df = pd.DataFrame({
        'velocity': velocities,
        'angle': angles,
        'range': measured_ranges
    })
    return df
```

```python
df = generate_dataset()
df.to_csv('simulation_dataset.csv', index=False)
df.head()
```
